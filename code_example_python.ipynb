```python
#1. Statistics for summarizing your sample
#import the numpy library as well as a function for computing IQR
import numpy as np
from scipy.stats import iqr
#define some variable
x=[1.3,0.9,1.1,3.2,1.6,1.7,0.8,1.0,1.4,1.5,1.3,1.5,1.7]
#print different statistics from this variable
print('Mean:',np.mean(x))
print('Standard deviation:',np.std(x))
print('Median:',np.median(x))
print('Interquartile range:',iqr(x))
print('Minimum:',np.min(x))
print('Maximum:',np.max(x))
print('Range:',np.max(x)-np.min(x))
```

```python
#2. Visualization
#Import matplotlib.pyplot library and a function for density plot
import matplotlib.pyplot as plt
from seaborn import kdeplot
#define a variable 
x=[1.3,0.9,1.1,3.2,1.6,1.7,0.8,1.0,1.4,1.5,1.3,1.5,1.7]
#plot a histogram, boxplot, and density plot
plt.hist(x,color='lightblue')
plt.show()
plt.boxplot(x)
plt.show()
kdeplot(x,color='blue')
plt.show()
```

```python
#3. Shapiro-Wilk normality test
from scipy.stats import shapiro
#define a variable
x=[1.3,0.9,1.1,3.2,1.6,1.7,0.8,1.0,1.4,1.5,1.3,1.5,1.7]
#print the output of the test for the variable
print(shapiro(x))
```

```python
#4. False discovery rate correction
from scipy.stats import false_discovery_control
#Suppose that you have the following p-values:
p_values=[0.03,0.00045,0.01,0.31,0.56,0.97,0.00001]
#print the corrected p-values
print('The corrected p-values:',false_discovery_control(p_values))
```

```python
#5. Paired and unpaired tests
from scipy.stats import ttest_rel, wilcoxon, ttest_ind, mannwhitneyu
#define two variables
x=[1.3,0.9,1.1,3.2,1.6,1.7,0.8,1.0,1.4,1.5,1.3,1.5,1.7]
y=[1.4,1.8,1.3,3.3,1.5,1.7,0.9,1.1,1.5,1.7,1.5,1.3,1.9]
print('Paired t-test:',ttest_rel(x,y))
print('Wilcoxon signed-rank test:',wilcoxon(x,y))
print('Unpaired t-test:',ttest_ind(x,y))
print('Mann-Whitney U test:',mannwhitneyu(x,y))
#To know which test to use, remember to check normality and consider if data is paired or not!
```

```python
#6. Tests comparing multiple groups
from scipy.stats import f_oneway, kruskal
#define three variables
x=[1.3,0.9,1.1,3.2,1.6,1.7,0.8,1.0,1.4,1.5,1.3,1.5,1.7]
y=[1.4,1.8,1.3,3.3,1.5,1.7,0.9,1.1,1.5,1.7,1.5,1.3,1.9]
z=[1.1,0.7,1.0,3.1,1.3,1.8,0.5,0.6,0.4,0.5,0.3,1.5,1.1]
print('ANOVA:',f_oneway(x,y,z))
print('Kruskal-Wallis test:',kruskal(x,y,z))
#To know which test to use, remember to check normality!
```

```python
#7. Correlation tests
from scipy.stats import pearsonr, spearmanr
x=[1.3,0.9,1.1,3.2,1.6,1.7,0.8,1.0,1.4,1.5,1.3,1.5,1.7]
y=[1.4,1.8,1.3,3.3,1.5,1.7,0.9,1.1,1.5,1.7,1.5,1.3,1.9]
print('Pearsons correlation:',pearsonr(x,y))
print('Spearmans correlation:',spearmanr(x,y))
#(Pearson's and Spearman's correlation coefficients are the statistics in the printed test output)
#To know which test to use, remember to check normality!
```

```python
#8. Tests for a contingency table
import numpy as np
from scipy.stats import chi2_contingency, fisher_exact
#Define a contingency table as numpy array
x=np.array([[31,89,97],[13,14,35]])
print(x)
#Print the output of the chi-squared test
print(chi2_contingency(x))
#Define another 2*2 table
y=np.array([[10,7],[3,1]])
print(y)
#Print the output of Fisher's exact test
print(fisher_exact(y))
```
